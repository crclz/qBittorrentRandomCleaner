```python
import requests
```

```python
# https://github.com/qbittorrent/qBittorrent/wiki/WebUI-API-(qBittorrent-5.0)
qb_url = "http://localhost:8199"
```

```python
response = requests.get(f"{qb_url}/api/v2/torrents/info")
response.text[:20]
```

```python
# get torrents
# status: download complete
response = requests.get(f"{qb_url}/api/v2/torrents/info", params={
    "filter": "completed",
})

len(response.json())
```

```python
from pathlib import Path
import time
# filter torrents
# download complete time: > 30 day
# in d://JellyShared folder

torrents = [p for p in response.json() if p["completion_on"] < int(time.time()) - 30 * 86400]
print(f"filter completion_on: {len(torrents)}")

def is_parent(parent, child):
    return parent.resolve() in Path(child).resolve().parents

folder_to_clean = Path("D://JellyShared").resolve()

torrents = [p for p in torrents if is_parent(folder_to_clean, p['content_path'])]
print(f"filter folder_to_clean: {len(torrents)}")

```

```python

import random

drop_prob = 0.3
drop_count = int(len(torrents) * drop_prob)

# sort torrents randomly
random.shuffle(torrents)

torrents_to_remove = torrents[:drop_count]

# list them from new to old
torrents_to_remove.sort(key=lambda p: p['completion_on'], reverse=True)

print(f"remove {len(torrents_to_remove)} torrents:")
for p in torrents_to_remove:
    print(p['content_path'])
```
